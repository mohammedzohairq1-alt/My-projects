```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("automobile_dataset.csv", encoding = "unicode escape")
```

```python
df.shape
```

```python
df.info()
```

## --> Start Data Cleaning

```python
df.isnull().sum()
```

```python
df.drop(["Accident_History"],axis = 1,inplace = True)
```

```python
df.head(20)
```

```python
int_blank = [
    "Engine_Size",
    "Horsepower",
    "Torque",
    "Fuel_Efficiency"
]

for bars in int_blank:
    df[bars] = df[bars].fillna(df[bars].median())
```

```python
str_blank = [
    "Transmission",
    "Service_History",
    "Color",
    "Location"
]
for bars in str_blank:
    df[bars] = df[bars].fillna(df[bars].mode()[0])
```

```python
df.head(20)
```

```python
df.info()
```

```python
df [["Selling_Price","Mileage","Fuel_Efficiency","Horsepower","Torque","Engine_Size"]].describe()
```

## -->Let's Start Analyse 

```python
plt.figure(figsize= (8,4))
sns.boxplot(x= df['Selling_Price'],color = 'lightgreen')
plt.title('Selling Price Outlier Detection')
plt.show()
```

```python
plt.figure(figsize = (8,4))
sns.boxplot(x = df['Mileage'], color = 'lightpink')
plt.title('Mileage Outlier Detection')
plt.show()
```

```python
# Cell 4: Both boxplots side by side
fig, axes = plt.subplots(1,2, figsize = (14,5))

sns.boxplot(x = df['Mileage'], ax= axes[0], color = 'lightpink')
axes[0].set_title('Mileage Outlier Detection')

sns.boxplot(x= df['Selling_Price'], ax= axes[1], color = 'lightgreen')
axes[1].set_title('Selling_Price Outlier Detection')

plt.tight_layout()
plt.show()
```

```python
Make_group = df.groupby(['Make'],as_index = False)['Selling_Price'].sum().sort_values(by = 'Selling_Price', ascending = False)
plt.figure(figsize = (12,5))
ax = sns.barplot(x ='Make', y='Selling_Price', data = Make_group)
for bars in ax.containers:
    ax.bar_label(bars, fmt = '%.0f', padding = 3)
plt.title('Company Wise Selling Price')
plt.show()
```

```python
Model_group = df.groupby(['Model'], as_index = False)['Selling_Price'].sum().sort_values(by = 'Selling_Price', ascending = False).head(10)
plt.figure(figsize = (12, 5))
ax = sns.barplot(x = 'Model', y = 'Selling_Price', data = Model_group)
for bars in ax.containers:
    ax.bar_label(bars, fmt = '%.0f', padding = 3)
plt.title('Model Wise Selling Price')
plt.show()
```

```python
Year_group = df.groupby(['Year'], as_index = False)['Selling_Price'].sum().sort_values(by = 'Selling_Price', ascending = False)
plt.figure(figsize = (12,5))
ax = sns.barplot(x = 'Year', y = 'Selling_Price', data = Year_group)
for bars in ax.containers:
    ax.bar_label(bars, fmt = '%.0f', padding = 3)
plt.title('Year Wise Selling Price')
plt.show()
```

```python
fuel_count = df['Fuel_Type'].value_counts()
plt.pie(
    fuel_count,
    labels = fuel_count.index,
    autopct = '%1.2f%%',
    startangle = 90
)
plt.title('Fuel Type Count')
plt.show()
```

```python
Engine_group = df.groupby(['Engine_Size'],as_index = False)['Selling_Price'].median().sort_values(by = 'Selling_Price', ascending = False).head(5)
plt.figure(figsize = (12,5))
ax = sns.barplot(x= 'Engine_Size', y= 'Selling_Price',data = Engine_group, order= Engine_group['Engine_Size'])
for bar in ax.containers:
    ax.bar_label(bar, fmt = '%.0f', padding = 3)
plt.title('Engine Size Wise Selling Price')
plt.show()
```

```python
df.columns
```

```python
plt.figure(figsize = (12,5))
df['Horsepower_Bin'] = pd.cut(df['Horsepower'], bins = [0, 100, 150, 200, 250, 300, 350, 400, 450],
                             labels = ['0-100', '101-150', '151-200', '201-250', '251-300', '301-350', '351-400', '401-450'])
Horsepower_group = df.groupby(['Horsepower_Bin'], as_index = False, observed = True)['Selling_Price'].mean()
ax = sns.barplot(data = Horsepower_group, x= 'Horsepower_Bin', y= 'Selling_Price')
for container in ax.containers:
    ax.bar_label(container, fmt = '%.0f', padding = 3)
plt.tight_layout()
plt.title('Horsepower Bins Average Selling Price')
plt.show()
```

```python
owner_count = df['Owners'].value_counts()
plt.pie(
    owner_count,
    labels = owner_count.index,
    autopct = '%1.2f%%',
    startangle = 90
)
plt.tight_layout()
plt.title('Owners Count')
plt.show()
```

```python
servicing_count = df['Service_History'].value_counts()
plt.pie(
    servicing_count,
    labels = servicing_count.index,
    autopct = '%1.2F%%',
    startangle = 90
)
plt.title('Service Count')
plt.show()
```

```python
Colors_count = df.groupby(['Color'], as_index = False)['Selling_Price'].sum().sort_values(by = 'Selling_Price', ascending = False)
plt.figure(figsize = (12,5))
ax = sns.barplot(data = Colors_count, x = 'Color', y = 'Selling_Price')
for bars in ax.containers:
    ax.bar_label(bars, fmt = '%.0f', padding=3)
plt.tight_layout()
plt.title('Colors Wise Selling Price')
plt.show()
```

```python
plt.figure(figsize = (12,5))
Fuel_Efficiency = df.groupby(['Fuel_Efficiency'], as_index = False)['Selling_Price'].mean().sort_values(by = 'Selling_Price', ascending = False).head(10)
ax = sns.barplot(data = Fuel_Efficiency, x = 'Fuel_Efficiency', y= 'Selling_Price')
for container in ax.containers:
    ax.bar_label(container, fmt = '%.0f', padding = 3)
plt.tight_layout()
plt.title('Fuel Efficiency')
plt.show()
```

```python
plt.figure(figsize = (12,5))
Body_Type = df.groupby(['Body_Type','Fuel_Type'], as_index = False)['Selling_Price'].sum()
ax = sns.barplot(data = Body_Type, x = 'Body_Type', y = 'Selling_Price', hue = 'Fuel_Type')
for container in ax.containers:
    ax.bar_label(container, fmt= '%.0f', padding = 3)
plt.tight_layout()
plt.title('Body Type Wise Selling')
plt.show()
```

```python
Drivetrain_Count = df['Drivetrain'].value_counts()
plt.pie(
    Drivetrain_Count,
    labels = Drivetrain_Count.index,
    autopct = '%1.2F%%',
    startangle = 90
)
plt.tight_layout()
plt.title('Drivetrain Count')
plt.show()
```
